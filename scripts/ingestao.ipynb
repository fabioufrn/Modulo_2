```python
import duckdb
import pandas as pd
import os
from datetime import datetime

# 1. Caminhos absolutos
caminho_banco = r"D:\Laboratorio\Projetos_Anhanguera_POS_GRAD\modulo_2\projeto\Modulo_2\dados_duckdb.db"
pasta_landing = r"D:\Laboratorio\Projetos_Anhanguera_POS_GRAD\modulo_2\projeto\Modulo_2\landing"
con = duckdb.connect(database=caminho_banco, read_only=False)

# 2. Lendo o PRIMEIRO arquivo
arquivo1 = 'z0019_1.csv'
df1 = pd.read_csv(os.path.join(pasta_landing, arquivo1), sep=';')
df1['nome_ingestao'] = arquivo1
df1['data_ingestao'] = datetime.now()

# 3. Lendo o SEGUNDO arquivo
arquivo2 = 'z0019_2.csv'
df2 = pd.read_csv(os.path.join(pasta_landing, arquivo2), sep=';')
df2['nome_ingestao'] = arquivo2
df2['data_ingestao'] = datetime.now()

# 4. Juntando (concatenando) os dois DataFrames em um só
df_final = pd.concat([df1, df2], ignore_index=True)

# 5. Apaga as tabelas antigas para não dar erro de duplicidade
con.execute("DROP TABLE IF EXISTS bronze_produtos")
con.execute("DROP TABLE IF EXISTS bronze_z0019")

# 6. Cria a tabela final no DuckDB com os 10 registros (5 de cada arquivo)
con.execute("CREATE TABLE bronze_z0019 AS SELECT * FROM df_final")

# 7. Salva e fecha a conexão
con.close()
print(f"Ingestão concluída! Total de registros salvos: {len(df_final)}")
```
